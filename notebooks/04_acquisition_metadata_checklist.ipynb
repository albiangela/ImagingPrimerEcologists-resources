# Acquisition Metadata Checklist

Good image analysis starts before recording. This notebook keeps the practical checklist close to the analysis notebooks: plan what must be measured, record enough metadata to reproduce the setup, and make file names machine-readable from the start.

The cells below create a lightweight metadata template you can adapt per study or session.


## Before recording

- Define the measurement target: color, position, speed, identity, body posture, scale, or habitat context.
- Decide whether color calibration, lens calibration, perspective rectification, or multi-camera synchronization is required.
- Record the reference target under the same lighting and camera settings as the experiment.
- Lock camera settings where possible: frame rate, resolution, exposure, white balance, focus, ISO/gain, and codec/profile.
- Use consistent file names without spaces, with zero-padded counters and ISO-style dates.
- Preserve raw files and metadata before making corrected derivatives.


```python
from datetime import date
import json

metadata_template = {
    "project": "example_project",
    "session_date": date.today().isoformat(),
    "site_or_lab": "",
    "operator": "",
    "camera": {
        "make_model": "",
        "lens": "",
        "resolution_px": [1920, 1080],
        "frame_rate_fps": 30,
        "exposure": "manual or auto setting",
        "white_balance": "fixed value or preset",
        "codec_or_format": "",
    },
    "calibration": {
        "color_reference": "gray card / ColorChecker / none",
        "lens_calibration": "calibration file or target used",
        "scale_reference": "object dimensions and units",
        "synchronization": "hardware trigger / software trigger / audio / none",
    },
    "lighting_environment": {
        "light_source": "",
        "diffusion_or_polarization": "",
        "notable_changes": "",
    },
    "file_naming": {
        "pattern": "YYYYMMDD_session##_camera##_take####.ext",
        "raw_storage_location": "",
        "derived_storage_location": "",
    },
}

print(json.dumps(metadata_template, indent=2))
```

## Quality-control pass

After recording a short test clip, check these before collecting the full dataset:

- Subject is visible, in focus, and large enough in pixels for the planned analysis.
- No important region is saturated, hidden by glare, or lost in shadows.
- Motion blur is acceptable for the behavior being measured.
- Calibration/reference targets are visible and measured.
- File names sort correctly and preserve session/camera/take identity.
- A raw backup exists before any conversion, trimming, compression, or correction.


```python
def make_filename(session_date, session, camera, take, suffix="mp4"):
    return f"{session_date}_session{session:02d}_camera{camera:02d}_take{take:04d}.{suffix}"


for take in range(1, 4):
    print(make_filename("2026-05-18", session=1, camera=2, take=take))
```

## What to keep

Keep raw files, calibration frames, reference-target dimensions, acquisition metadata, and any notebook versions used to generate derivatives. Corrected videos, rectified frames, and plots can be regenerated, so they should usually live in ignored output folders rather than version control.
